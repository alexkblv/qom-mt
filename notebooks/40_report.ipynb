Rebuilds the experiment's tables, inline-number macros and machine-readable summary from
the artifacts on disk, under `artifacts/tables/<experiment>/`.

Nothing here is typed by hand: the metric column header is derived from the metric's own
parameters, split sizes come from split manifests, and any evaluation recorded as
contaminated is excluded from the results tables rather than quietly included.

```python
EXPERIMENT = "baseline"
```

```python
from qomlaq.pipeline import report

written = report(EXPERIMENT)
for name, path in written.items():
    print(f"{name:28s} {path}")
```
