# 10 · Train one run

Fine-tunes one run of an experiment: one split in one direction, starting from the
experiment's model profile.

The corpus and splits are rebuilt from the raw sources first, and the split is loaded by
the hash the experiment pins. Rebuilding is deterministic and takes seconds, and it means
a run cannot be handed a split built from different sources.

**No metrics are computed here.** Scoring happens in `20_evaluate.ipynb` against the saved
checkpoint, so the released artifact and the evaluated artifact are the same object, and
re-scoring never requires retraining.

```python
EXPERIMENT = "baseline"
RUN_ID = "base__stratified__es2qom"     # `qomlaq runs baseline` lists them
```

```python
from qomlaq.pipeline import train

card = train(EXPERIMENT, RUN_ID)
print("selection:", card.selection["criterion"],
      "| best epoch:", card.selection["best_epoch"],
      "| epochs run:", card.selection["epochs_run"])
card.training["truncation"]
```
