# 01 · Build the splits

Builds the experiment's corpus (if it is not already built) and every split the experiment
declares, partitioned at the discourse-unit level.

Every split is checked before it is written: units, exact pairs and normalized duplicates
must not straddle partitions, held-out probes must stay held out, and Bible verses must
not sit next to a training verse. A headline split failing any of these is not written.
Each split is then compared with the hash its experiment file pins, and a mismatch stops
the build.

```python
EXPERIMENT = "baseline"
```

```python
from qomlaq.pipeline import build

built = build(EXPERIMENT)
```

## Check results

`bible_verse_adjacency` is the share of held-out verses whose neighboring verse is in
train. The verse-level control should show ~0.94; every headline split 0.0.

```python
for split_id, manifest in built.manifests.items():
    print(f"\n=== {split_id}  [{manifest['purpose']}]")
    for name, outcome in manifest["checks"].items():
        mark = "pass" if outcome["passed"] else "FAIL"
        print(f"  [{mark}] {name}: {outcome['value']}")
```
